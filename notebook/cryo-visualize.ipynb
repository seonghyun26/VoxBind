# EMDB Density Map + CrossDocked Pocket Visualization

Visualize cryo-EM density maps alongside X-ray pocket–ligand pairs from the CrossDocked dataset.
The mapping follows the chain: **CrossDocked PDB → UniProt → EM PDB entries → EMDB accession → `.map.gz`**

| Section | Description |
|---|---|
| **Load Data** | CrossDocked test set + EMDB mapping |
| **Select Sample** | Pick a pocket / EMDB map by index |
| **Load EMDB Map** | Decompress & subsample the cryo-EM density |
| **PDB Structure** | Cartoon + ligand via nglview |
| **Side-by-side** | Pocket atoms (left) vs. cryo-EM isosurface (right) |
| **Contour sweep** | Adjust iso-percentile to show more/less density |
| **Browse** | Summary table of all paired samples |

```python
import os, sys, gzip, json, tempfile, warnings
import numpy as np
import torch
from pathlib import Path

import plotly.graph_objects as go
from plotly.subplots import make_subplots
import mrcfile
import nglview as nv

warnings.filterwarnings('ignore')

NOTEBOOK_DIR = Path(os.getcwd())
REPO_ROOT    = NOTEBOOK_DIR.parent
DATA_DIR     = REPO_ROOT / "voxbind" / "dataset" / "data"
OUT_DIR      = NOTEBOOK_DIR / "data"
EMDB_DIR     = OUT_DIR / "emdb"

POCKET_CH = {0: 'C', 1: 'O', 2: 'N', 3: 'S'}
LIGAND_CH = {0: 'C', 1: 'O', 2: 'N', 3: 'S', 4: 'F', 5: 'Cl', 6: 'P'}
ATOM_COLOR = {
    'C': '#909090', 'O': '#FF3030', 'N': '#3050F8', 'S': '#E0C020',
    'F': '#33FF33', 'Cl': '#1FF01F', 'P': '#FF8000',
}
ATOM_VDW = {'C': 1.70, 'O': 1.52, 'N': 1.55, 'S': 1.80, 'F': 1.47, 'Cl': 1.75, 'P': 1.80}
POCKET_RADII_LUT = np.array([ATOM_VDW[POCKET_CH[i]] for i in range(4)], dtype=np.float32)
LIGAND_RADII_LUT = np.array([ATOM_VDW[LIGAND_CH[i]] for i in range(7)], dtype=np.float32)

print(f"Data: {DATA_DIR}\nEMDB: {EMDB_DIR}")
```

## 1. Load Data

Loads the CrossDocked test split and the pre-built `crossdocked_pdb_to_emdb.json` mapping, then filters to samples that have at least one downloaded EMDB `.map.gz`.

```python
def extract_pdb_id(pocket_id: str) -> str:
    """'FAMILY/4abc_A_rec_..._pocket10.pdb' -> '4ABC'"""
    return pocket_id.split("/")[1].split("_")[0].upper()

def emdb_num(emdb_id: str) -> str:
    """'EMD-5778' -> '5778'"""
    return emdb_id.split("-")[-1]

def find_local_maps(emdb_ids: list) -> list:
    return [EMDB_DIR / f"emd_{emdb_num(e)}.map.gz"
            for e in emdb_ids
            if (EMDB_DIR / f"emd_{emdb_num(e)}.map.gz").exists()]

# Load test split
print("Loading data_test.pt ...")
samples = torch.load(DATA_DIR / "data_test.pt", weights_only=False)

# Load EMDB mapping built by the pipeline notebook
with open(OUT_DIR / "crossdocked_pdb_to_emdb.json") as f:
    cd_to_emdb: dict = json.load(f)

# Find test samples that have at least one downloaded EMDB map
paired = []
for pocket, ligand in samples:
    pid      = extract_pdb_id(pocket["id"])
    emdb_ids = cd_to_emdb.get(pid, [])
    maps     = find_local_maps(emdb_ids)
    if maps:
        paired.append(dict(
            pocket=pocket, ligand=ligand,
            pdb_id=pid, emdb_ids=emdb_ids, local_maps=maps
        ))

print(f"Test samples         : {len(samples)}")
print(f"With local EMDB maps : {len(paired)}")
print()
for p in paired[:8]:
    print(f"  {p['pdb_id']}  →  {p['emdb_ids'][:2]}  ({len(p['local_maps'])} local map(s))")
```

## 2. Select Sample

Set `IDX` to pick a pocket from the `paired` list and `EMDB_IDX` to choose which EMDB map to display (some PDB IDs map to multiple).

```python
IDX      = 0   # index into `paired`
EMDB_IDX = 0   # which EMDB map to display

sample   = paired[IDX]
pocket   = sample["pocket"]
ligand   = sample["ligand"]
pdb_id   = sample["pdb_id"]
emdb_id  = sample["emdb_ids"][EMDB_IDX]
map_path = sample["local_maps"][min(EMDB_IDX, len(sample["local_maps"]) - 1)]

# Raw .pt data: pocket has keys {id, coords, atoms_channel}
#              ligand has keys  {id, coords, atoms_channel, max_len}
# No radius key — filter by valid atom channel and look up VdW radii.
p_mask = pocket["atoms_channel"].numpy() < 4   # C/O/N/S only
l_mask = ligand["atoms_channel"].numpy()  < 7   # C/O/N/S/F/Cl/P only

p_coords = pocket["coords"].numpy()[p_mask]
p_ch     = pocket["atoms_channel"].numpy()[p_mask]
p_radii  = POCKET_RADII_LUT[p_ch]

l_coords = ligand["coords"].numpy()[l_mask]
l_ch     = ligand["atoms_channel"].numpy()[l_mask]
l_radii  = LIGAND_RADII_LUT[l_ch]

print(f"PDB ID    : {pdb_id}")
print(f"EMDB ID   : {emdb_id}")
print(f"Map       : {map_path.name}  ({map_path.stat().st_size / 1e6:.0f} MB)")
print(f"Pocket ID : {pocket['id']}")
print()
print(f"Pocket atoms : {p_coords.shape[0]}")
print(f"Ligand atoms : {l_coords.shape[0]}")
print(f"Pocket centre (approx.) : {p_coords.mean(0).round(1)}")
print(f"Ligand centre           : {l_coords.mean(0).round(1)}")
```

## 3. Load EMDB Map

Decompresses the `.map.gz`, reads it with `mrcfile`, and subsamples to `MAP_TARGET` voxels per side for fast rendering. Percentile-based iso-contour thresholds are computed from the density distribution.

```python
def scatter_atoms(coords, channels, radii, ch_map, opacity=0.8, size_scale=4.0, name_prefix=""):
    """Return one Scatter3d trace per element type (for use with add_trace)."""
    traces = []
    for ch_idx in np.unique(channels):
        elem = ch_map.get(int(ch_idx), f"ch{int(ch_idx)}")
        mask = channels == ch_idx
        size = float(np.clip(np.median(radii[mask]) * size_scale, 2, 12))
        traces.append(go.Scatter3d(
            x=coords[mask, 0], y=coords[mask, 1], z=coords[mask, 2],
            mode="markers",
            marker=dict(size=size, color=ATOM_COLOR.get(elem, "#888"), opacity=opacity),
            name=f"{name_prefix} {elem}",
            legendgroup=f"{name_prefix}_{elem}",
        ))
    return traces


def load_mrc_gz(path: Path, target_dim: int = 80):
    """Decompress .map.gz, read with mrcfile, subsample. Returns (data, res_ang, step)."""
    with gzip.open(path, "rb") as gz:
        raw = gz.read()
    with tempfile.NamedTemporaryFile(suffix=".map", delete=False) as tmp:
        tmp.write(raw)
        tmp_path = tmp.name
    try:
        with mrcfile.open(tmp_path, mode="r", permissive=True) as mrc:
            data      = mrc.data.copy().astype(np.float32)
            voxel_ang = float(mrc.voxel_size.x)
    finally:
        os.unlink(tmp_path)
    step = max(1, data.shape[0] // target_dim)
    sub  = data[::step, ::step, ::step]
    return sub, voxel_ang * step, step
```

```python
MAP_TARGET = 80   # target voxels per side (lower → faster render)

print(f"Loading {map_path.name} ({map_path.stat().st_size/1e6:.0f} MB) ...")
sub, res, step = load_mrc_gz(map_path, MAP_TARGET)

D, H, W = sub.shape
lin = lambda n: (np.arange(n) - n / 2) * res
Z, Y, X = np.meshgrid(lin(D), lin(H), lin(W), indexing="ij")

iso_lo = float(np.percentile(sub, 99.5))
iso_hi = float(np.percentile(sub, 99.97))

print(f"Shape: ~{sub.shape[0]*step}³  →  {sub.shape}  (stride ×{step},  {res:.2f} Å/voxel)")
print(f"Contour: {iso_lo:.4f} – {iso_hi:.4f}")
```

## 4. PDB Structure — nglview

Interactive cartoon view of the CrossDocked X-ray structure with the small-molecule ligand (ball+stick) and a faint protein surface.

```python
view = nv.show_pdbid(pdb_id, height="500px", width="800px")
view.clear_representations()
view.add_representation("cartoon",    color="chainid")
view.add_representation("ball+stick", selection="ligand",  color="element")
view.add_representation("surface",    selection="protein",
                         opacity=0.08, color="white")
view.center()
view
```

## 5. Side-by-side: Pocket Atoms vs. EMDB Density

Left panel: pocket (grey, transparent) + ligand (coloured by element) in the X-ray coordinate frame.  
Right panel: cryo-EM isosurface of the same protein family (linked via UniProt). Coordinate frames differ — direct overlay requires alignment.

```python
fig_combined = make_subplots(
    rows=1, cols=2,
    specs=[[{"type": "scene"}, {"type": "scene"}]],
    subplot_titles=(
        f"Pocket + Ligand  ({pdb_id}, X-ray)",
        f"cryo-EM density  ({emdb_id})",
    ),
    horizontal_spacing=0.02,
)

# ── Left panel: pocket + ligand ──────────────────────────────────────────────
for tr in scatter_atoms(p_coords, p_ch, p_radii, POCKET_CH,
                         opacity=0.18, size_scale=3.0, name_prefix="Pocket"):
    fig_combined.add_trace(tr, row=1, col=1)

for tr in scatter_atoms(l_coords, l_ch, l_radii, LIGAND_CH,
                         opacity=0.95, size_scale=5.5, name_prefix="Ligand"):
    fig_combined.add_trace(tr, row=1, col=1)

# ── Right panel: EMDB isosurface ─────────────────────────────────────────────
fig_combined.add_trace(
    go.Isosurface(
        x=X.flatten(), y=Y.flatten(), z=Z.flatten(),
        value=sub.flatten(),
        isomin=iso_lo, isomax=iso_hi,
        surface_count=3,
        opacity=0.55,
        colorscale="Blues",
        showscale=False,
        caps=dict(x_show=False, y_show=False, z_show=False),
        name=emdb_id,
    ),
    row=1, col=2,
)

fig_combined.update_layout(
    title_text=f"{pdb_id}  ↔  {emdb_id}   (same protein family via UniProt)",
    scene=dict(
        aspectmode="data",
        bgcolor="rgb(15,17,22)",
        xaxis_showbackground=False,
        yaxis_showbackground=False,
        zaxis_showbackground=False,
    ),
    scene2=dict(
        aspectmode="data",
        xaxis=dict(showbackground=False),
        yaxis=dict(showbackground=False),
        zaxis=dict(showbackground=False),
    ),
    legend=dict(groupclick="toggleitem"),
    width=1400, height=700,
    paper_bgcolor="white",
)
fig_combined.show()
```

## 6. Contour Sensitivity Sweep

Adjust `pct_lo` / `pct_hi` to expose more or less of the density envelope.  
`pct_lo ≈ 99.0` shows the full protein shell; `pct_lo ≈ 99.9` shows only the densest core.

```python
pct_lo = 99.2   # lower → more density shown
pct_hi = 99.98  # upper bound on the colour scale

iso_lo2 = float(np.percentile(sub, pct_lo))
iso_hi2 = float(np.percentile(sub, pct_hi))
print(f"Contour {pct_lo}–{pct_hi}th percentile: {iso_lo2:.4f} – {iso_hi2:.4f}")

fig_c = go.Figure(go.Isosurface(
    x=X.flatten(), y=Y.flatten(), z=Z.flatten(),
    value=sub.flatten(),
    isomin=iso_lo2, isomax=iso_hi2,
    surface_count=4,
    opacity=0.45,
    colorscale="Teal",
    showscale=True,
    caps=dict(x_show=False, y_show=False, z_show=False),
))
fig_c.update_layout(
    title=f"{emdb_id}  |  contour {pct_lo}–{pct_hi} percentile",
    scene=dict(
        xaxis=dict(title="x (Å)", showbackground=False),
        yaxis=dict(title="y (Å)", showbackground=False),
        zaxis=dict(title="z (Å)", showbackground=False),
        aspectmode="data",
    ),
    width=850, height=680,
)
fig_c.show()
```

## 7. Browse All Paired Samples

Summary table of every test-set pocket that has at least one downloaded EMDB map. Use `idx` values to set `IDX` in the selection cell above.

```python
rows = []
for i, p in enumerate(paired):
    pk = p["pocket"]
    lg = p["ligand"]
    rows.append({
        "idx"         : i,
        "pdb_id"      : p["pdb_id"],
        "pocket_atoms": int((pk["atoms_channel"].numpy() < 4).sum()),
        "ligand_atoms": int((lg["atoms_channel"].numpy() < 7).sum()),
        "n_emdb_ids"  : len(p["emdb_ids"]),
        "n_local_maps": len(p["local_maps"]),
        "emdb_ids"    : ", ".join(p["emdb_ids"][:3]),
    })

import pandas as pd
df = pd.DataFrame(rows)
df
```
